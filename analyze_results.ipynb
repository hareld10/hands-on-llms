```python
metric_sums = {}
data = df.metrics.values
for entry in data:
    for key, value in entry.items():
        metric_sums[key] = metric_sums.get(key, 0) + value

# Calculate means
mean_metrics = {key: value / len(data) for key, value in metric_sums.items()}

# Print the mean of each metric
print("Mean Metrics:")
for metric, mean_value in mean_metrics.items():
    print(f"{metric}: {mean_value:.4f}")
```
